# Model 0

```python
!pip install biopython
from Bio.PDB import PDBParser
import pandas as pd

pdb_file = "/content/pkhd1.pdb"  # Replace with your actual PDB path
parser = PDBParser(QUIET=True)
structure = parser.get_structure("model", pdb_file)

data = []

# Standard 3-letter to 1-letter amino acid mapping
aa_3to1 = {
    'ALA':'A', 'CYS':'C', 'ASP':'D', 'GLU':'E', 'PHE':'F',
    'GLY':'G', 'HIS':'H', 'ILE':'I', 'LYS':'K', 'LEU':'L',
    'MET':'M', 'ASN':'N', 'PRO':'P', 'GLN':'Q', 'ARG':'R',
    'SER':'S', 'THR':'T', 'VAL':'V', 'TRP':'W', 'TYR':'Y'
}

for model in structure:
    for chain in model:
        for residue in chain:
            if residue.id[0] != ' ':  # Skip non-standard residues
                continue
            res_name_3 = residue.get_resname()
            res_name_1 = aa_3to1.get(res_name_3, 'X')  # Fallback to 'X' for unknowns
            res_id = residue.id[1]
            plddt = list(residue.get_atoms())[0].get_bfactor()  # pLDDT from first atom
            data.append([res_id, res_name_1, plddt])

# Convert to DataFrame and save
df = pd.DataFrame(data, columns=["Residue_Index", "Residue", "pLDDT"])
df.to_csv("residue_plddt_scores.csv", index=False)
average_plddt = df["pLDDT"].mean()
print("Overall average pLDDT score:", average_plddt)
print("CSV file saved: residue_plddt_scores.csv")
```

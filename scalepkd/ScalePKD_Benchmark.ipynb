# **ClinVar txt to csv file**

```python
import pandas as pd
import re
from google.colab import files
from IPython.display import FileLink

# Step 1: Load the file (adjust delimiter as needed)
df = pd.read_csv('/content/ClinVar_PKHD1_Benign.txt', delimiter='\t')

# Step 2: Amino acid dictionary (3-letter to 1-letter)
aa_dict = {
    'Ala': 'A', 'Arg': 'R', 'Asn': 'N', 'Asp': 'D', 'Cys': 'C',
    'Glu': 'E', 'Gln': 'Q', 'Gly': 'G', 'His': 'H', 'Ile': 'I',
    'Leu': 'L', 'Lys': 'K', 'Met': 'M', 'Phe': 'F', 'Pro': 'P',
    'Ser': 'S', 'Thr': 'T', 'Trp': 'W', 'Tyr': 'Y', 'Val': 'V'
}

# Step 3: Convert 'p.Ala345Pro' or 'p.(Ala345Pro)' → 'A345P'
def convert_col1_to_1letter(entry):
    entry = str(entry).strip()
    match = re.search(r"p\.?\(?([A-Z][a-z]{2})(\d+)([A-Z][a-z]{2})\)?", entry)
    if match:
        from_aa, pos, to_aa = match.groups()
        return f"{aa_dict.get(from_aa, '?')}{pos}{aa_dict.get(to_aa, '?')}"
    return None

# Step 4: Apply conversion
df['AA_1letter_from_col1'] = df.iloc[:, 0].apply(convert_col1_to_1letter)

# Step 5: Drop rows where conversion failed
df = df.dropna(subset=['AA_1letter_from_col1'])

# Step 6: Extract amino acid position as integer
df['AA_position'] = df['AA_1letter_from_col1'].str.extract(r'[A-Z](\d+)[A-Z]')[0].astype(int)

# Step 7: Save intermediate file
converted_filename = "/content/converted_from_column1.csv"
df.to_csv(converted_filename, index=False)
print("Download your file with the new column:")
display(FileLink(converted_filename))

# Step 8: Drop rows missing required fields
required_cols = ['VariationID', 'Gene(s)', 'Protein change', 'Germline classification']
df = df.dropna(subset=required_cols)

# Step 9: Remove duplicated Variant IDs
df = df[~df['VariationID'].duplicated(keep=False)].copy()

# Step 10: Sort by amino acid position (and optionally gene)
sorted_data = df.sort_values(by=['Gene(s)', 'AA_position'])  # use ['AA_position'] if you don't want gene grouping

# Step 11: Save final sorted data
sorted_filename = '/content/sorted_PKHD1_benign_ClinVar.csv'
sorted_data.to_csv(sorted_filename, index=False)

# Step 12: Print classification counts
classification_counts = sorted_data['Germline classification'].value_counts()
print("\n✅ File processed successfully!\n")
print("Classification counts:\n")
print(classification_counts)

# Step 13: Download final file
print("\nDownload sorted final file:")
display(FileLink(sorted_filename))
```

# **Add ScalePKD to the dataset**

```python
import re
import numpy as np
import pandas as pd

# ============================
# Files
# ============================
SCALEPKD_FILE = "ScalePKD (1).csv"
PATH_FILE     = "/content/ClinVar_PKHD1_Pathogenic.txt"
BENIGN_FILE   = "/content/ClinVar_PKHD1_Benign.txt"

# ============================
# Helpers
# ============================
AA3_TO_AA1 = {
    "Ala":"A","Arg":"R","Asn":"N","Asp":"D","Cys":"C","Gln":"Q","Glu":"E","Gly":"G",
    "His":"H","Ile":"I","Leu":"L","Lys":"K","Met":"M","Phe":"F","Pro":"P","Ser":"S",
    "Thr":"T","Trp":"W","Tyr":"Y","Val":"V","Ter":"*","Stop":"*"
}

def clean_cols(df):
    df.columns = (
        df.columns
          .str.strip()
          .str.lower()
          .str.replace(" ", "_")
    )
    return df

def normalize(v):
    if pd.isna(v):
        return np.nan

    s = str(v).replace("p.", "").replace(" ", "")

    # 1-letter format (V3433M)
    m = re.match(r"^([A-Z\*])(\d+)([A-Z\*])$", s)
    if m:
        return s

    # 3-letter format (Val3433Met)
    m = re.match(r"^([A-Z][a-z]{2})(\d+)([A-Z][a-z]{2}|Ter|\*)$", s)
    if m:
        return (
            AA3_TO_AA1[m.group(1)]
            + m.group(2)
            + AA3_TO_AA1.get(m.group(3), "*")
        )

    return np.nan

# ============================
# Load ClinVar
# ============================
path = clean_cols(pd.read_csv(PATH_FILE, sep="\t"))
ben  = clean_cols(pd.read_csv(BENIGN_FILE, sep="\t"))

path["truth"] = 1
ben["truth"]  = 0

df = pd.concat([path, ben], ignore_index=True)

# ============================
# Normalize ClinVar variants
# ============================
df["variant_1l"] = df["protein_change"].apply(normalize)
df = (
    df.dropna(subset=["variant_1l"])
      .drop_duplicates("variant_1l")
)

# ============================
# Load ScalePKD (correct file)
# ============================
scale = clean_cols(pd.read_csv(SCALEPKD_FILE))
scale["variant_1l"] = scale["protein_variant"].apply(normalize)

# ============================
# Merge
# ============================
m = df.merge(
    scale[["variant_1l", "scalepkd_prediction"]],
    on="variant_1l",
    how="inner"
)

# ============================
# Predictions
# ============================
m["pred"] = (
    m["scalepkd_prediction"]
      .str.lower()
      .str.strip()
      .eq("pathogenic")
      .astype(int)
)

# ============================
# Confusion matrix
# ============================
TP = ((m.truth == 1) & (m.pred == 1)).sum()
FN = ((m.truth == 1) & (m.pred == 0)).sum()
TN = ((m.truth == 0) & (m.pred == 0)).sum()
FP = ((m.truth == 0) & (m.pred == 1)).sum()

# ============================
# Metrics
# ============================
accuracy    = (TP + TN) / (TP + TN + FP + FN)
sensitivity = TP / (TP + FN) if TP + FN else 0
specificity = TN / (TN + FP) if TN + FP else 0
precision   = TP / (TP + FP) if TP + FP else 0

# ============================
# Output
# ============================
print("\n=== ScalePKD performance (PKHD1) ===")
print(f"Variants evaluated : {len(m)}")
print(f"Accuracy           : {accuracy:.3f}")
print(f"Sensitivity (TPR)  : {sensitivity:.3f}")
print(f"Specificity (TNR)  : {specificity:.3f}")
print(f"Precision (PPV)    : {precision:.3f}")
print(f"TP={TP}, FN={FN}, TN={TN}, FP={FP}")
```

```python
import numpy as np

def metrics(df):
    TP = ((df.truth == 1) & (df.pred == 1)).sum()
    FN = ((df.truth == 1) & (df.pred == 0)).sum()
    TN = ((df.truth == 0) & (df.pred == 0)).sum()
    FP = ((df.truth == 0) & (df.pred == 1)).sum()

    return {
        "accuracy":    (TP + TN) / (TP + TN + FP + FN),
        "sensitivity": TP / (TP + FN) if TP + FN else np.nan,
        "specificity": TN / (TN + FP) if TN + FP else np.nan,
        "precision":   TP / (TP + FP) if TP + FP else np.nan
    }

# ----------------------------
# Bootstrap
# ----------------------------
B = 10000
boot = {k: [] for k in metrics(m).keys()}

for _ in range(B):
    samp = m.sample(len(m), replace=True)
    for k, v in metrics(samp).items():
        boot[k].append(v)

# ----------------------------
# Report 95% CIs
# ----------------------------
print("\n=== 95% bootstrap confidence intervals ===")
for k, vals in boot.items():
    lo, hi = np.percentile(vals, [2.5, 97.5])
    print(f"{k:12s}: {np.mean(vals):.3f} (95% CI {lo:.3f}–{hi:.3f})")
```

# **AlphaMissense comparison**

```python
import pandas as pd
import numpy as np
import re

# ============================
# Files
# ============================
import pandas as pd
import numpy as np
import re

# ============================
# Files
# ============================
SCALEPKD_FILE = "ScalePKD (1).csv"
PATH_FILE     = "/content/sorted_PKHD1_pathogenic_ClinVar.csv"
BENIGN_FILE   = "/content/sorted_PKHD1_benign_ClinVar.csv"

# ============================
# Helpers
# ============================
AA3_TO_AA1 = {
    "Ala":"A","Arg":"R","Asn":"N","Asp":"D","Cys":"C","Gln":"Q","Glu":"E","Gly":"G",
    "His":"H","Ile":"I","Leu":"L","Lys":"K","Met":"M","Phe":"F","Pro":"P","Ser":"S",
    "Thr":"T","Trp":"W","Tyr":"Y","Val":"V","Ter":"*","Stop":"*"
}

def clean_cols(df):
    df.columns = (
        df.columns
          .str.strip()
          .str.lower()
          .str.replace(" ", "_")
    )
    return df

def normalize(v):
    if pd.isna(v):
        return np.nan
    s = str(v).replace("p.", "").replace(" ", "")
    m = re.match(r"^([A-Z\*])(\d+)([A-Z\*])$", s)
    if m:
        return s
    m = re.match(r"^([A-Z][a-z]{2})(\d+)([A-Z][a-z]{2}|Ter|\*)$", s)
    if m:
        return f"{AA3_TO_AA1[m.group(1)]}{m.group(2)}{AA3_TO_AA1.get(m.group(3),'*')}"
    return np.nan

# ============================
# Load truth sets (ClinVar)
# ============================
path = clean_cols(pd.read_csv(PATH_FILE))
ben  = clean_cols(pd.read_csv(BENIGN_FILE))

path["truth"] = 1
ben["truth"]  = 0

df = pd.concat([path, ben], ignore_index=True)

# ClinVar variant column (known + correct)
VAR_COL = "protein_change"
if VAR_COL not in df.columns:
    raise KeyError(f"Expected '{VAR_COL}' in ClinVar files, found:\n{df.columns.tolist()}")

df["variant_1l"] = df[VAR_COL].apply(normalize)
df = df.dropna(subset=["variant_1l"]).drop_duplicates("variant_1l")

# ============================
# Load ScalePKD (AlphaMissense)
# ============================
scale = clean_cols(pd.read_csv(SCALEPKD_FILE))

if "protein_variant" not in scale.columns:
    raise KeyError(f"'protein_variant' not found in ScalePKD columns:\n{scale.columns.tolist()}")

scale["variant_1l"] = scale["protein_variant"].apply(normalize)

# ============================
# Merge AlphaMissense
# ============================
m = df.merge(
    scale[["variant_1l", "pathogenicity_class"]],
    on="variant_1l",
    how="inner"
)

# ============================
# AlphaMissense prediction logic
# ============================
CLASS_MAP = {
    "likely_pathogenic": 1,
    "likely_benign": 0,
    "ambiguous": -1
}

m["pred"] = (
    m["pathogenicity_class"]
      .str.lower()
      .str.strip()
      .map(CLASS_MAP)
)

if m["pred"].isna().any():
    raise ValueError("Unexpected AlphaMissense class encountered")

# ============================
# Confusion matrix (ambiguous = wrong)
# ============================
TP = ((m.truth == 1) & (m.pred == 1)).sum()
TN = ((m.truth == 0) & (m.pred == 0)).sum()

FN = (
    ((m.truth == 1) & (m.pred == 0)) |
    ((m.truth == 1) & (m.pred == -1))
).sum()

FP = (
    ((m.truth == 0) & (m.pred == 1)) |
    ((m.truth == 0) & (m.pred == -1))
).sum()

# ============================
# Metrics
# ============================
accuracy    = (TP + TN) / (TP + TN + FP + FN)
sensitivity = TP / (TP + FN)
specificity = TN / (TN + FP)
precision   = TP / (TP + FP)

print("\n=== AlphaMissense performance (ambiguous counted as wrong) ===")
print(f"Variants evaluated : {len(m)}")
print(f"Accuracy           : {accuracy:.3f}")
print(f"Sensitivity        : {sensitivity:.3f}")
print(f"Specificity        : {specificity:.3f}")
print(f"Precision          : {precision:.3f}")
print(f"TP={TP}, FN={FN}, TN={TN}, FP={FP}")
print(f"Ambiguous counted as errors: {(m.pred == -1).sum()}")

# ============================
# Bootstrap 95% confidence intervals
# ============================
def metrics(df):
    TP = ((df.truth == 1) & (df.pred == 1)).sum()
    TN = ((df.truth == 0) & (df.pred == 0)).sum()

    FN = (
        ((df.truth == 1) & (df.pred == 0)) |
        ((df.truth == 1) & (df.pred == -1))
    ).sum()

    FP = (
        ((df.truth == 0) & (df.pred == 1)) |
        ((df.truth == 0) & (df.pred == -1))
    ).sum()

    return {
        "accuracy":    (TP + TN) / (TP + TN + FP + FN),
        "sensitivity": TP / (TP + FN),
        "specificity": TN / (TN + FP),
        "precision":   TP / (TP + FP)
    }

B = 10000
boot = {k: [] for k in metrics(m).keys()}

for _ in range(B):
    samp = m.sample(len(m), replace=True)
    for k, v in metrics(samp).items():
        boot[k].append(v)

print("\n=== AlphaMissense 95% bootstrap CIs (ambiguous = wrong) ===")
for k, vals in boot.items():
    lo, hi = np.percentile(vals, [2.5, 97.5])
    print(f"{k:12s}: {np.mean(vals):.3f} (95% CI {lo:.3f}–{hi:.3f})")
```
